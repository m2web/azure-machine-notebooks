```python
from azure.core.credentials import AzureKeyCredential
from azure.search.documents.indexes import SearchIndexerClient
from azure.search.documents.indexes.models import (
    SearchIndexerDataSourceConnection,
    SearchIndexerDataContainer
)

# 1. Initialize Client
indexer_client = SearchIndexerClient(
    endpoint="https://rush-lyric-ai-search.search.windows.net", 
    credential=AzureKeyCredential("<REDACTED_API_KEY>")
)

# 2. Define Data Source
data_source_connection = SearchIndexerDataSourceConnection(
    name="rush-lyrics-ds",
    type="azureblob",
    connection_string="<REDACTED_CONNECTION_STRING>",
    container=SearchIndexerDataContainer(name="rush-lyrics-data")
)

# 3. Create Connection
indexer_client.create_data_source_connection(data_source_connection)
```
