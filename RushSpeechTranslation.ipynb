Translate iconic Rush lyrics into Japanese (JA) and French (FR) in real-time.

```python
import azure.cognitiveservices.speech as speechsdk
```

```python
API_KEY = "******"
REGION = "eastus"

translation_config = speechsdk.translation.SpeechTranslationConfig(subscription=API_KEY, region=REGION)
translation_config.speech_recognition_language = "en-US"
translation_config.add_target_language("ja") # Japanese
translation_config.add_target_language("fr") # French

audio_config = speechsdk.audio.AudioConfig(filename="rush_lyrics.wav")
translator = speechsdk.translation.TranslationRecognizer(translation_config=translation_config, audio_config=audio_config)
```

## Translation Test
Try speaking the chorus of 'Tom Sawyer':
**"A modern-day warrior, mean, mean stride, today's Tom Sawyer, he gets high on you."**

```python
print("Speak Rush lyrics in English...")
result = translator.recognize_once_async().get()

if result.reason == speechsdk.ResultReason.TranslatedSpeech:
    print(f"To Japanese: {result.translations['ja']}")
    print(f"To French: {result.translations['fr']}")
else:
    print(f"Error: {result.reason}")
```
