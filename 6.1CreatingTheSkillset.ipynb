```python
from azure.core.credentials import AzureKeyCredential
from azure.search.documents.indexes import SearchIndexerClient
from azure.search.documents.indexes.models import (
    AzureOpenAIEmbeddingSkill,
    InputFieldMappingEntry,
    OutputFieldMappingEntry,
    SearchIndexerSkillset
)
```

```python
# 1. Initialize Client
indexer_client = SearchIndexerClient(
    endpoint="https://rush-lyric-ai-search.search.windows.net", 
    credential=AzureKeyCredential("<REDACTED_API_KEY>")
)
```

```python
# 2. Define the Embedding Skill
embedding_skill = AzureOpenAIEmbeddingSkill(
    description="Compute embeddings for the content field",
    context="/document",
    resource_url="https://temp-m2web-ai-search-pr-resource.cognitiveservices.azure.com/",
    deployment_name="text-embedding-3-small",
    model_name="text-embedding-3-small",
    api_key="<REDACTED_API_KEY>",
    inputs=[InputFieldMappingEntry(name="text", source="/document/content")],
    outputs=[OutputFieldMappingEntry(name="embedding", target_name="vector")]
)
```

```python
# 3. Create the Skillset object
skillset = SearchIndexerSkillset(
    name="rush-lyrics-skillset",
    skills=[embedding_skill],
    description="Skillset to automatically embed Rush lyrics during ingestion"
)
```

```python
indexer_client.create_skillset(skillset)
```
