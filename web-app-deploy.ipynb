```python
from openai import AzureOpenAI
```

```python
endpoint = "https://m2-ai-102-openai.cognitiveservices.azure.com/openai/deployments/gpt-4o-mini/chat/completions?api-version=2025-01-01-preview"
deployment_name = "gpt-4o-mini"
api_key = "<REDACTED_API_KEY>"
```

```python
client = AzureOpenAI(
    azure_endpoint=endpoint,
    api_key=api_key,
    api_version="2024-07-01-preview"
)

try:
    print(f"Sending request to model: {deployment_name}...")
    completion = client.chat.completions.create(
        model=deployment_name,
        messages=[
            {
                "role": "user",
                "content": "What is your favorite Rush album?",
            }
        ],
    )

    print("\nResponse:")
    if completion.choices:
        print(completion.choices[0].message.content)
    else:
        print("No choices returned. Full response:")
        print(completion)

except Exception as e:
    print(f"\nAN ERROR OCCURRED:\n{e}")
```
