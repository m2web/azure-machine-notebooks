```python
import os
from openai import AzureOpenAI
import json
```

```python
endpoint = "https://m2web-mjakqnon-swedencentral.cognitiveservices.azure.com/openai/deployments/dall-e-3/images/generations?api-version=2024-02-01"
apikey = "<REDACTED_API_KEY>"
```

```python
openai.api_key = apikey
```

```python
client = AzureOpenAI(
    api_version="2024-05-01-preview",
    azure_endpoint=endpoint,
    api_key=apikey,
)

result = client.images.generate(
    model="dall-e-3", # the name of your DALL-E 3 deployment
    prompt="Geddy Lee, Alex Lifeson, and Anika Nilles of the Canadian power trio (with Anika from Germany) on stage in the Rush 2026 Reunion tour.",
    n=1
)

image_url = json.loads(result.model_dump_json())['data'][0]['url']

print(image_url)
```

```python
import requests
from datetime import datetime
# Download and save the image
response = requests.get(image_url)
if response.status_code == 200:
    # Generate a unique filename using the current timestamp
    timestamp = datetime.now().strftime("%Y%m%d_%H%M%S")
    filename = f"generated_image_{timestamp}.png"
    
    # Save the image in the current directory
    with open(filename, 'wb') as f:
        f.write(response.content)
    print(f"Image saved as {filename}")
else:
    print("Failed to download the image")
```
