### Importing Libraries and Setting Up Client

#### Add the following Python imports to your notebook

```python
from azure.ai.documentintelligence import DocumentIntelligenceClient
from azure.core.credentials import AzureKeyCredential
import os
```

```python
endpoint = "********"
key = "********"
```

```python
# We specify the latest GA API version
client = DocumentIntelligenceClient(
    endpoint=endpoint, 
    credential=AzureKeyCredential(key),
    api_version="2024-11-30"
)
```

```python
# Replace with your actual filename
path_to_receipt = "invoice_v1_1_standard.pdf"

with open(path_to_receipt, "rb") as f:
    poller = client.begin_analyze_document(
        "prebuilt-receipt", 
        body=f,
        content_type="application/octet-stream"
    )
    result = poller.result()
```

```python
for document in result.documents:
    fields = document.fields
    
    merchant = fields.get('MerchantName')
    date = fields.get('TransactionDate')
    total = fields.get('Total')

    print(f"Merchant Name: {merchant.value_string if merchant else 'N/A'}")
    print(f"Transaction Date: {date.value_date if date else 'N/A'}")
    print(f"Total Amount: {total.value_currency.amount if total else 'N/A'}")
```
