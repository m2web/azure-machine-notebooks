```python
from azure.ai.contentsafety import ContentSafetyClient
from azure.core.credentials import AzureKeyCredential

API_KEY = "*************"
ENDPOINT = "https://temp-rush-speech.cognitiveservices.azure.com/"
```

```python
# Initialize the client
client = ContentSafetyClient(endpoint=ENDPOINT, credential=AzureKeyCredential(API_KEY))
```

#### Analyze Text for Harmful Content

```python
from azure.ai.contentsafety.models import AnalyzeTextOptions

text_input = "The night is black, without a moon. The air is thick and still. The vigilantes gather on the staircase to the hill. Burning eyes of hatred and ill-will. Madmen fed on fear and lies. To beat, and burn, and kill."
request = AnalyzeTextOptions(text=text_input)

try:
    response = client.analyze_text(request)
    print(f"Analysis for: '{text_input}'")
    for item in response.categories_analysis:
        print(f"{item.category} severity: {item.severity}")
except Exception as e:
    print(f"Error: {e}")
```

#### Analyze Images for Harmful Content

```python
import os
from azure.ai.contentsafety.models import AnalyzeImageOptions, ImageData

image_path = "starman.png"

if os.path.exists(image_path):
    with open(image_path, "rb") as image_data:
        request = AnalyzeImageOptions(image=ImageData(content=image_data.read()))
        
        try:
            response = client.analyze_image(request)
            print(f"Analysis Results for {image_path}:")
            for item in response.categories_analysis:
                print(f"{item.category} severity: {item.severity}")
        except Exception as e:
            print(f"Error: {e}")
else:
    print(f"File {image_path} not found. Please upload the Starman image.")
```
