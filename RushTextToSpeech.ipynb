This notebook uses Azure AI Speech to synthesize iconic lyrics from the band **Rush** using expressive **SSML**.

```python
import os
import azure.cognitiveservices.speech as speechsdk
```

```python
API_KEY = "*****"
REGION = "eastus"

speech_config = speechsdk.SpeechConfig(subscription=API_KEY, region=REGION)
audio_config = speechsdk.audio.AudioOutputConfig(use_default_speaker=True)
```

## Synthesizing 'The Spirit of Radio'
We use the `cheerful` style for these optimistic lyrics.

```python
ssml_text = """
<speak version='1.0' xmlns='http://www.w3.org/2001/10/synthesis' xmlns:mstts='https://www.w3.org/2001/mstts' xml:lang='en-US'>
    <voice name='en-US-JennyNeural'>
        <mstts:express-as style='cheerful'>
            Begin the day with a friendly voice. 
            <break time='300ms'/>
            A companion unobtrusive, plays that song that's so elusive.
            <break time='500ms'/>
            And the magic music makes your morning mood.
        </mstts:express-as>
    </voice>
</speak>
"""

speech_synthesizer = speechsdk.SpeechSynthesizer(speech_config=speech_config, audio_config=audio_config)
result = speech_synthesizer.speak_ssml_async(ssml_text).get()
```

```python
if result.reason == speechsdk.ResultReason.SynthesizingAudioCompleted:
    print("Rush lyrics synthesis completed.")
else:
    print(f"Error: {result.reason}")
```
